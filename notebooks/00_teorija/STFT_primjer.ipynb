```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import stft

fs = 16000

# trajanje
t = np.linspace(0, 1.5, int(fs*1.5))

signal = np.zeros_like(t)

# 0-0.5s : 200 Hz
signal[:fs//2] = np.sin(2*np.pi*200*t[:fs//2])

# 0.5-1s : 800 Hz
signal[fs//2:fs] = np.sin(2*np.pi*800*t[fs//2:fs])

# 1-1.5s : 2000 Hz
signal[fs:] = np.sin(2*np.pi*2000*t[fs:])

# STFT
f, time, Zxx = stft(signal,fs,nperseg=512)


plt.figure(figsize=(12,5))
plt.pcolormesh(time,f,np.abs(Zxx),shading='gouraud')
plt.ylabel("Frekvencija [Hz]")
plt.xlabel("Vrijeme [s]")
plt.title("STFT spektrogram primjer")
plt.ylim(0,3000)
plt.colorbar(label="Energija")
plt.show()
```
